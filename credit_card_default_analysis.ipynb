# Credit Card Default Analysis — SQL Project



## Project Overview

This project analyzes credit card client data using Python, pandas, SQLite, and SQL. The main objective is to explore the relationship between client characteristics, credit limits, repayment history, and credit card default.
## Key Findings

###1. Overall Default Patterns

The dataset contains 30,000 clients. The analysis examines the number and percentage of clients who defaulted, using Y = 1 to identify default.

###2. Education

Default rates varied across education categories. Among the most common categories, category 3 had a higher default rate than categories 1 and 2. Results for less common categories should be interpreted cautiously because their sample sizes were small.

###3. Credit Limit

Clients in lower credit-limit groups generally had higher default rates. The default rate ranged from 31.79% for the 0–50,000 group to 13.26% for the 300,001+ group.

###4. Repayment Status

Clients with positive repayment-status codes in September had higher default rates, particularly when the recorded code was 2 or greater. Some categories contained relatively few clients.

###5. Number of Months with Positive Delay Codes

0 months: 11.71% default rate

1–2 months: 32.51% default rate

3+ months: 59.99% default rate

The default rate increased as the number of months with positive delay codes increased.

###6. Age

The relationship between age and default was not linear. The youngest and oldest age groups had higher default rates than some middle-age groups.

###7. Cross-Analysis

The relationship between repayment history and default remained visible across age groups and credit-limit groups. Clients with positive delay codes in three or more months had substantially higher default rates in every group examined.

## Conclusion

The strongest pattern observed in this analysis was the association between repeated positive repayment-status codes and credit card default. Credit limit, education, and age also showed differences in default rates, but these relationships were less uniform.

These findings describe associations in the dataset and do not establish causation. Positive repayment-status codes are used here as indicators for grouping clients; they should not be interpreted as a precise measure of total days or months overdue.
